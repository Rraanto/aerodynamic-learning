# Project description 

The goal of this project to apply functional learning methods to learn entire aerodynamic function representations. For each airfoil parameterised by some finite-dimensional vector $K \in \mathbb R^k$, learn $C_D(K, \alpha, v), C_L(K, \alpha, v)$ to quickly learn the tendency of the airfoil without much computation. This is typically known as an angle or velocity sweep. Different methods can be thought of, starting with some decomposition (neural operators) or multidimensional learning. 

This can be thought of as a field learning as well, if we consider immediately angle of attack and velocity at the same time. 

The perspectives of this is to add functional descriptors directly into optimisation formulations, eventually allowing for faster exploration; and allowing computationally cheap formulations targetting objectives such as "optimise low-Reynolds and zero angle drag" costing only one model call (Geometry to aerodynamic). Currently, this is somewhat cheaply done using neuralfoil, which itself is already an extremely cheap surrogate (trained on XFoil, itself somewhat low-fidelity). 

## Project plan 
This project will, to some extent, not be valuable unless some actual insight on aerodynamic _functionals_ are made, thus it will be relevant to start by analysing those said functionals on a given family of airfoils. 

A rough initial project plan can be defined as follows: 

- Problem setup: choose a given family of airfoils, fidelity level, operational conditions to target 
- Elementary Analysis: Analyse how the aerodynamic functionals of those airfoils evolve. This initial analysis is ideally a simple visually-assisted analysis (basically run analysis on different airfoils), trying to spot geometric descriptors for example and physically justify them when possible. 
- Further Analysis: Explore different algorithmic analysis methods (I mostly think about functional decompositions, descriptors, ...) that can eventually be learned or used as learning targets, optimisation criterias, ... etc This part can particularly end with methods like clustering methods (functional clustering? ... )
- Training a model as a synthesis: Choose some descriptors / high dimensional / functional data representation and train a model to compute those representations. The first analogy / idea that comes to my mind is some sort of Fourier decomposition of functional outputs turning the problem into learning the decomposition coefficients (analogous to (?) neural operators)
- Using the trained model in an optimisation problem: Should the efficiency of the trained model be justified, this can turn into an epistemic search of "nice formulations", in any case it can/should be very informative. 

Initially, I think going up to the "further analysis" part can already be rich in information. Using low-fidelity screening tools (notably Neuralfoil), it is fairly reasonable to believe that a very vast exploration can be done quickly. 

Benchmarking is something that can be done systematically from further analysis on: different descriptors can be benchmarked 

## Tools 
The basic project will start on the analysis package that Aerosandbox offers: 
- Aerosandbox models (VLM, ...)
- neuralfoil (screening analysis)
- XFoil 

Those should give both rapidity for quick analysis and scaling perspectives (from 2D to 3D, from NF to XF,...) but at very advanced stages, things like porting the pipeline to a higher-fidelity (CFD, ...) can be imagined. The pipeline is moreover generally modulable as it needs the following model types : 
- One model type for "manual" analysis (ideally very fast, somehow accurate) 
- One model type for "further analysis", ideally model outputs are somehow representative of the fidelity levels that we want in an optimisation loop, 
- One model type for actually training a new model (for example, if some decomposition in some basis has been deemed efficient, use that approach in a CFD high fidelity database )
