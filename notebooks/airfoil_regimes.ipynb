# Airfoil geometry and operating regimes

Does the global functional 2D map preserve contour neighbors, and how do performance clusters change across nine angle-magnitude/Reynolds regimes? This extension reuses the saved 170-airfoil fields. See `airfoil_regimes.md` for assumptions and `../report/airfoil_regimes.pdf` for the report.


```python
from pathlib import Path
import runpy
import pandas as pd
from IPython.display import Image, display
root = Path.cwd()
if not (root / "utils/airfoil_regimes.py").exists():
    root = root.parent
assert (root / "utils/airfoil_regimes.py").exists()
runpy.run_path(str(root / "notebooks/airfoil_regimes.py"), run_name="__main__")
data = root / "data/airfoil_regimes"
figures = root / "figures/airfoil_regimes"
```

## Geometry on the existing projected points
Registered contour RMS uses upper/lower surface coordinates and 201 stations. Neighbor overlap is descriptive; pair distances are not independent observations.

## Nine operating regimes and performance clusters
Angle magnitude bins are [0,5], [5,10], [10,15] degrees; Reynolds bins divide [1,2] million equally. These are operating bins, not stall labels. Performance uses drag, signed lift, absolute lift, and mean pointwise absolute-lift/drag. Clustering uses 27 standardized drag, absolute-lift, and efficiency descriptors.


```python
display(pd.read_csv(data / "cluster_geometry.csv"))
display(pd.read_csv(data / "cluster_performance.csv"))
for name in ["regime_clusters", "performance"]:
    display(Image(filename=str(figures / f"{name}.png")))
```

## Local fields, reconstruction, and sensitivity
Each local partition uses fixed global coefficient scales. Local fits optimize their own metric; this is an in-population descriptive comparison. Colors in local partition panels are not aligned. Inspect sensitivity before interpreting cluster labels as physical classes.
